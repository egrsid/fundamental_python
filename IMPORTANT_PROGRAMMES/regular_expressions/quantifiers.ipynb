## Квантификаторы {m, n}
m - минимальное число совпадений с выражением

n - максимальное число совпадений с выражением

```python
s = 'google gooogle goooogle'
```

```python
# мажорный квантификатор - берет наибольшую возможную длину строки
pattern = r'o{2,5}'  # находит комбинации "o" от 2 до 5
```

```python
# минорный квантификатор - берет наименьшую возможную длину строки
pattern = r'o{2,5}?'
```

```python
pattern = r'go{2}gle'  # повторение ровно 2 раза
```

```python
pattern = r'go{2,}gle'  # повторение от 2 раз
```

```python
pattern = r'go{,4}gle'  # повторение до 4 раз
```

? - аналог {0,1}

\* - аналог {0,}

\+ - аналог {1,}

```python
s = 'автор=Пушкин А.С;title = Евгений Онегин;year=   2001'

pattern = r'\w+\s*=\s*[^;]*'  # разбивает по парам ключ-значение
```

```python
s = "<p>Картинка <img src = 'bg_color.jpg'> в тексте </p>"

pattern = r'<img\s+[^>]*?src\s*=\s*[^>]+'  # выделяет <img src = 'bg_color.jpg'> из строки
```

```python
match = re.findall(pattern, s)
match
```
