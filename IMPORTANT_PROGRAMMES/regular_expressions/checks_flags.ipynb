```python
import re
```

```python
s = 'подоходный налог'
```

```python
pattern = r"прибыль|обретение|\bдоход\b"  # ставим \b, чтобы искалось только слово целиком
```

```python
pattern = r"\b(?:прибыль|обретение|доход)\b"  # чтобы не делать проверку каждого условия, а делать это сруз для группы
```

```python
match = re.findall(pattern, s)
match
```
